```python
#Data
import pandas as pd

#Plots
import matplotlib.pyplot as plt

#math
import numpy as np
```

```python
Snap_number=5260
df_1=np.loadtxt('Runs/Run_1/snaps/'+f'snapsh_{Snap_number}.dat', skiprows=3)
df_2=np.loadtxt('Runs/run2/snaps/'+f'snapsh_{Snap_number}.dat', skiprows=3)
df_3=np.loadtxt('Runs/Run_3/snaps/'+f'snapsh_{Snap_number}.dat', skiprows=3)
df_4=np.loadtxt('Runs/Run4/snaps/'+f'snapsh_{Snap_number}.dat', skiprows=3)
df_5=np.loadtxt('Runs/Run_zhou/snaps/'+f'snapsh_{Snap_number}.dat', skiprows=3)
df_6=np.loadtxt('Runs/Run_zhoureddi/snaps/'+f'snapsh_{Snap_number}.dat', skiprows=3)
df_7=np.loadtxt('Runs/Run_tian/snaps/'+f'snapsh_{Snap_number}.dat', skiprows=3)
df_8=np.loadtxt('Runs/Run_tianreddi/snaps/'+f'snapsh_{Snap_number}.dat', skiprows=3)
```

```python

def read_abundance_file_A(df:list):
    A=[0]
    abundances=[0]
    for line in df:
        values = list(map(float, line))
        A_=values[0]+values[1]
        if A_>A[-1]:
            A.append(A_)
            abundances.append(values[3])#X
        else:
            i=np.searchsorted(A, A_)
            abundances[i]+=values[3]
    return A, abundances
A_1,Y_1=read_abundance_file_A(df_1)
A_2,Y_2=read_abundance_file_A(df_2)
A_3,Y_3=read_abundance_file_A(df_3)
A_4,Y_4=read_abundance_file_A(df_4)
A_5,Y_5=read_abundance_file_A(df_5)
A_6,Y_6=read_abundance_file_A(df_6)
A_7,Y_7=read_abundance_file_A(df_7)
A_8,Y_8=read_abundance_file_A(df_8)
```

```python
plt.rcParams.update({
    "font.size": 12,
    "axes.labelsize": 14,
    "legend.fontsize": 12,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12
})

plt.figure(figsize=(10,4), dpi=300)

plt.scatter(A_1,Y_1,label=r'Run1',s=7)
#plt.scatter(A_2,Y_2,label=r'Run2',s=4)
#plt.scatter(A_3,Y_3,label=r'Run3',color='C3',s=5)
#plt.scatter(A_4,Y_4,label=r'Run4',color='C5',s=5)
#plt.scatter(A_5,Y_5,label=r'Run5',color='C6',s=5)
#plt.scatter(A_6,Y_6,label=r'Run6',color='C2',s=5,marker='x')
plt.scatter(A_7,Y_7,label=r'Run7',color='C8',s=5)
plt.scatter(A_8,Y_8,label=r'Run8',color='C9',s=5)

plt.yscale('log')
plt.xlabel('A')
plt.ylabel(r'Fraction number density  $Y_i=\frac{n_i}{n}$')
plt.legend(loc="upper right", ncol=2,frameon=False,fontsize=8)
plt.ylim(1e-20,1.5)
#plt.xlim(0.1,500)
plt.savefig("Run1vsRun7vs8_abundance.pdf", bbox_inches="tight")
```
