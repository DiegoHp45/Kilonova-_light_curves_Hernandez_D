```python
from pathlib import Path
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('Empirical_beta_decay_reddi.csv')
name_p=data['name_p']
name_d=data['name_d']
z=data['Z']
n=data['N']
lam=data['decay rate (1/s)']
Q_value=data['Q_value']
```

```python
output_file=Path('Empirical_beta_decay_reddi_R1')
with output_file.open('w')as fout:
    fout.write('1'+' '*73 + '\n')
    fout.write(' '*74 + '\n')
    fout.write(' '*74 + '\n')
    for i in range(len(data)):
        fout.write(' '*(10-len(name_p[i]))+name_p[i]+' '*(5-len(name_d[i]))+name_d[i]+' '*28+'wc12w'+' '*5+f'{Q_value[i]:.5e}'+'          '+'\n')
        lamd=np.log(lam[i])
        if lamd>0:
            fout.write(' '+f'{lamd:.6e}'+' 0.000000e+00 0.000000e+00 0.000000e+00                      '+'\n')
        else:
            fout.write(f'{lamd:.6e}'+' 0.000000e+00 0.000000e+00 0.000000e+00                      '+'\n')
        fout.write(' 0.000000e+00 0.000000e+00 0.000000e+00                                   '+'\n')

```
