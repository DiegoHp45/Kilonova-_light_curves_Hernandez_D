```python
import numpy as np
import matplotlib.pyplot as plt
import os
import pynucastro as pyna
import pandas as pd
#%matplotlib widget
```

```python
approach_I='epsilon_calculations/approach_I/'
approach_II='epsilon_calculations/approach_II/'
runs_I=['Run_1', 'Run_2', 'Run_3', 'Run_4','Run_4_3', 'Run_5','Run_5_3','Run_10', 'Run_10_3', 'Run_11', 'Run_12']
runs_II=['Run_1', 'Run_2', 'Run_Tian', 'Run_Tian_NMHF','Run_Tian_NMHF_3', 'Run_Tian_royer','Run_Tian_SONG', 'Run_Tian_SONG_3', 'Run_Zhou_NMHF', 'Run_Zhou_royer', 'Run_Zhou_SONG', 'Run_Zhou_NMHF_3']
results=['/a_top', '/b_top']
folder=approach_II+runs_II[0]+results[0]
files=os.listdir(folder)
path=os.path.join(folder,files[0])
```

```python
#alpha
nucleos_a_I=[]
nucleos_a_II=[]
for run in runs_II:
    folder=approach_II+run+results[0]
    files=os.listdir(folder)
    for file in files:
        nuc=file.split('_')
        if nuc[0] not in nucleos_a_II:
            nucleos_a_II.append(nuc[0])
for run in runs_I:
    folder=approach_I+run+results[0]
    files=os.listdir(folder)
    for file in files:
        nuc=file.split('_')
        if nuc[0] not in nucleos_a_I:
            nucleos_a_I.append(nuc[0])
print('Vector de nucleos alpha_I relevantas son:'+str(np.sort(nucleos_a_I))+'\n n='+str(len(nucleos_a_I)))
print('Vector de nucleos alpha_II relevantes son:'+str(np.sort(nucleos_a_II))+'\n n='+str(len(nucleos_a_II)))
```

```python
#beta
nucleos_b_I=[]
nucleos_b_II=[]
for run in runs_II:
    folder=approach_II+run+results[1]
    files=os.listdir(folder)
    for file in files:
        nuc=file.split('_')
        if nuc[0] not in nucleos_b_II:
            nucleos_b_II.append(nuc[0])
for run in runs_I:
    folder=approach_I+run+results[1]
    files=os.listdir(folder)
    for file in files:
        nuc=file.split('_')
        if nuc[0] not in nucleos_b_I:
            nucleos_b_I.append(nuc[0])
print('Vector de nucleos beta_I relevantas son:'+str(np.sort(nucleos_b_I))+'\n n='+str(len(nucleos_b_I)))
print('Vector de nucleos beta_II relevantes son:'+str(np.sort(nucleos_b_II))+'\n n='+str(len(nucleos_b_II)))
```

```python
N_a_I=[]
Z_a_I=[]
for nucleo_a_I in nucleos_a_I:
    nucleo_a_I=pyna.Nucleus(nucleo_a_I)
    N_a_I.append(nucleo_a_I.N)
    Z_a_I.append(nucleo_a_I.Z)

N_a_II=[]
Z_a_II=[]
for nucleo_a_II in nucleos_a_II:
    nucleo_a_II=pyna.Nucleus(nucleo_a_II)
    N_a_II.append(nucleo_a_II.N)
    Z_a_II.append(nucleo_a_II.Z)

N_b_I=[]
Z_b_I=[]
for nucleo_b_I in nucleos_b_I:
    nucleo_b_I=pyna.Nucleus(nucleo_b_I)
    N_b_I.append(nucleo_b_I.N)
    Z_b_I.append(nucleo_b_I.Z)

N_b_II=[]
Z_b_II=[]
for nucleo_b_II in nucleos_b_II:
    nucleo_b_II=pyna.Nucleus(nucleo_b_II)
    N_b_II.append(nucleo_b_II.N)
    Z_b_II.append(nucleo_b_II.Z)
```

```python
nuclei_in_the_reaction=pd.read_csv(r'Nuclear_Data/Original_files/winvne_v2.0.dat',skiprows=lambda x: not (x>7854 and (x-7855)%4==0),delim_whitespace=True,names=['name','A','Z','N','spin','Mass excess (Mev)','source'])

plt.rcParams.update({
    "font.size": 12,
    "axes.labelsize": 17,
    "legend.fontsize": 20,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12
})

fig, ax = plt.subplots(1,1,figsize=(16,8))

ax.set_xlabel("N")
ax.set_ylabel("Z")
ax.set_xlim(-0.5, 240.5)
ax.set_ylim(-0.5, 116.5)
ax.set_aspect('equal')
ax.scatter(nuclei_in_the_reaction['N'], nuclei_in_the_reaction['Z'], color='lightgray', label='Nucleos en la red',marker='s')
#ax.scatter(N_a_I, Z_a_I, color='blue', label='Nucleos alpha_I',marker='s')
ax.scatter(N_a_II, Z_a_II, color='red', label='Nucleos alpha_II',marker='s')
#ax.scatter(N_b_I, Z_b_I, color='cyan', label='Nucleos beta_I',marker='s')
#ax.scatter(N_b_II, Z_b_II, color='magenta', label='Nucleos beta_II',marker='s')
ax.legend(loc='upper right')
```

```python
initial_time=1.0006912859 #seconds
final_time=4.32E7 #seconds 500 days= 4.32E7 seconds 100 days= 8.64E7 seconds
N_steps=10000
time = np.exp(np.linspace(np.log(initial_time), np.log(final_time), N_steps))
time_days=time/(24*60*60)
```
