```python
import pynucastro as pyna
```

```python
rates=pyna.rates.library.Library('Reaclib_exp_R1')
#beta decay 
filter_beta=pyna.RateFilter(max_reactants=1,
                            max_products=1,

                            filter_function=lambda r:  r.reactants[0].Z+1==r.products[0].Z and r.reactants[0].A==r.products[0].A
                            )

#alpha decay 
filter_alpha=pyna.RateFilter(products=['he4'],
                             exact=False,
                             max_reactants=1,
                             max_products=2,
                            
                             filter_function=lambda r: r.reactants[0].Z==r.products[1].Z+r.products[0].Z and r.reactants[0].A==r.products[1].A+r.products[0].A
                             )

beta=rates.filter(filter_beta)
alpha=rates.filter(filter_alpha)
print(len(beta.get_rates()),len(alpha.get_rates()))
```
