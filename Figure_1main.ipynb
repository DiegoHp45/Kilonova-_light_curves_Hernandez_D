```python

#Data
import pandas as pd

#Plots
import matplotlib.pyplot as plt

#math
import numpy as np


#%matplotlib widget
```

```python

df_1 = pd.read_csv("Runs/Run_1/generated_energy.dat",sep="\s{2,}",skiprows=2,header=None, engine='python')
df_2=pd.read_csv("Runs/run2/generated_energy.dat",sep="\s{2,}",skiprows=2,header=None,engine='python')
df_3=pd.read_csv("Runs/Run_3/generated_energy.dat",sep='\s{2,}',skiprows=2,header=None, engine= 'python')
df_4=pd.read_csv("Runs/Run4/generated_energy.dat",sep='\s{2,}',skiprows=2,header=None, engine= 'python')
df_5=pd.read_csv("Runs/Run_zhou/generated_energy.dat",sep='\s{2,}',skiprows=2,header=None, engine= 'python')
df_7=pd.read_csv("Runs/Run_tian/generated_energy.dat",sep='\s{2,}',skiprows=2,header=None, engine= 'python')
df_6=pd.read_csv("Runs/Run_zhoureddi/generated_energy.dat",sep='\s{2,}',skiprows=2,header=None, engine= 'python')
df_8=pd.read_csv("Runs/Run_tianreddi/generated_energy.dat",sep='\s{2,}',skiprows=2,header=None, engine= 'python')
```

```python
'''Preliminar figures'''
plt.rcParams.update({
    "font.size": 12,
    "axes.labelsize": 14,
    "legend.fontsize": 12,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12
})

plt.figure(figsize=(6.5,5.5), dpi=300)

plt.plot(df_1[0]/(24*60*60),df_1[1],label=r'Run1(Reaclib)')
plt.plot(df_2[0]/(24*60*60),df_2[1],label=r'Run2(Reaclib)',linestyle='--')
plt.plot(df_3[0]/(24*60*60),df_3[1],label=r'Run3(Exp)')
plt.plot(df_4[0]/(24*60*60),df_4[1],label=r'Run4(Exp)',linestyle='--')
plt.plot(df_5[0]/(24*60*60), df_5[1], label=r'Run5(Zhou)')
plt.plot(df_6[0]/(24*60*60), df_6[1], label=r'Run6(ZhouReddi)', linestyle='--')
plt.plot(df_7[0]/(24*60*60), df_7[1], label=r'Run7(Tian)')
plt.plot(df_8[0]/(24*60*60), df_8[1], label=r'Run8(TianReddi)', linestyle='--')

plt.yscale('log')
plt.xscale('log')
plt.xlabel('Time[Days]')
plt.ylabel(r'Effective heating rate $[erg\,g^{-1}\,s^{-1}]$')
plt.legend(loc="upper right", ncol=2,frameon=False,fontsize=8)
plt.ylim(1e4,1e14)
plt.xlim(0.1,500)
plt.savefig("Epsilon_Approach_II_runs_1_8.pdf", bbox_inches="tight")
```
