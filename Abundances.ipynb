```python

# ── Nuclear abundance trajectory tools ────────────────────────────────────────
import os, glob, re
from io import StringIO

# Z → chemical-symbol lookup (Z=0 = neutron by convention)
_Z_TO_SYM = {
    0:'n',  1:'H',  2:'He', 3:'Li', 4:'Be', 5:'B',  6:'C',  7:'N',  8:'O',  9:'F',
   10:'Ne',11:'Na',12:'Mg',13:'Al',14:'Si',15:'P', 16:'S', 17:'Cl',18:'Ar',19:'K',
   20:'Ca',21:'Sc',22:'Ti',23:'V', 24:'Cr',25:'Mn',26:'Fe',27:'Co',28:'Ni',29:'Cu',
   30:'Zn',31:'Ga',32:'Ge',33:'As',34:'Se',35:'Br',36:'Kr',37:'Rb',38:'Sr',39:'Y',
   40:'Zr',41:'Nb',42:'Mo',43:'Tc',44:'Ru',45:'Rh',46:'Pd',47:'Ag',48:'Cd',49:'In',
   50:'Sn',51:'Sb',52:'Te',53:'I', 54:'Xe',55:'Cs',56:'Ba',57:'La',58:'Ce',59:'Pr',
   60:'Nd',61:'Pm',62:'Sm',63:'Eu',64:'Gd',65:'Tb',66:'Dy',67:'Ho',68:'Er',69:'Tm',
   70:'Yb',71:'Lu',72:'Hf',73:'Ta',74:'W', 75:'Re',76:'Os',77:'Ir',78:'Pt',79:'Au',
   80:'Hg',81:'Tl',82:'Pb',83:'Bi',84:'Po',85:'At',86:'Rn',87:'Fr',88:'Ra',89:'Ac',
   90:'Th',91:'Pa',92:'U', 93:'Np',94:'Pu',95:'Am',96:'Cm',97:'Bk',98:'Cf',99:'Es',
  100:'Fm',101:'Md',102:'No',103:'Lr',104:'Rf',105:'Db',106:'Sg',107:'Bh',108:'Hs',
  109:'Mt',110:'Ds',111:'Rg',112:'Cn',
}
_SYM_TO_Z = {sym.lower(): Z for Z, sym in _Z_TO_SYM.items()}

_SIM_CACHE = {}   # run_id → DataFrame


def parse_snapshot(path):
    """Parse one WinNet snapshot file.

    Returns (time [s], T9 [GK], rho [g/cm³], df)
    where df has columns [N, Z, A, Y, X].
    """
    with open(path) as fh:
        lines = fh.readlines()

    # line 0: "    time    temp    dens"   (header – skip)
    # line 1: actual time / T9 / rho values
    vals = lines[1].split()
    time = float(vals[0])
    T9   = float(vals[1])
    rho  = float(vals[2])
    # line 2: "     nin     zin       y       x"  (column header – skip)
    # lines 3+: nucleus data
    data_str = ''.join(lines[3:])
    df = pd.read_csv(
        StringIO(data_str),
        sep=r'\s+', header=None,
        names=['N', 'Z', 'Y', 'X'],
        dtype={'N': int, 'Z': int, 'Y': float, 'X': float},
    )
    df['A'] = df['N'] + df['Z']
    return time, T9, rho, df


def load_simulation(run_id, base="Runs"):
    """Load all snapshots for a simulation run into a long-format DataFrame.

    Parameters
    ----------
    run_id : str
        Run identifier, e.g. '1', 'Tian', 'Zhou_NMHF'.
    base : str
        Path to the Runs root directory (default: 'Runs').

    Returns
    -------
    pandas.DataFrame
        MultiIndex (snap_idx, N, Z).
        Columns: time [s], T9 [GK], rho [g/cm³], A, Y, X.
        snap_idx=0 is the earliest snapshot; files are sorted numerically.
    """
    if run_id in _SIM_CACHE:
        return _SIM_CACHE[run_id]

    snap_dir = os.path.join(base, f"Run_{run_id}", "snaps")
    files = glob.glob(os.path.join(snap_dir, "snapsh_*.dat"))
    files.sort(key=lambda p: int(re.search(r'snapsh_(\d+)\.dat', os.path.basename(p)).group(1)))

    if not files:
        raise FileNotFoundError(f"No snapshots found in {snap_dir!r}")

    chunks = []
    for snap_idx, path in enumerate(files):
        time, T9, rho, df = parse_snapshot(path)
        df['snap_idx'] = snap_idx
        df['time'] = time
        df['T9']   = T9
        df['rho']  = rho
        chunks.append(df)

    sim = pd.concat(chunks, ignore_index=True)
    sim = sim.set_index(['snap_idx', 'N', 'Z'])
    _SIM_CACHE[run_id] = sim
    return sim


# ── Nucleus specification helpers ─────────────────────────────────────────────

def _parse_nucleus(spec):
    """Convert a nucleus spec to (Z, A).

    Accepted forms:
      (Z, A) tuple  →  passed through
      'n'  /  'neutron'  →  (0, 1)
      'p'  /  'proton'   →  (1, 1)
      'He-4', 'Fe56', 'Ni-78', …
    """
    if isinstance(spec, tuple):
        return spec
    s = spec.strip()
    if s.lower() in ('n', 'neutron'):
        return (0, 1)
    if s.lower() in ('p', 'proton'):
        return (1, 1)
    m = re.match(r'([A-Za-z]+)-?(\d+)$', s)
    if m:
        sym, A = m.group(1), int(m.group(2))
        Z = _SYM_TO_Z.get(sym.lower())
        if Z is None:
            raise ValueError(f"Unknown element symbol: {sym!r}")
        return (Z, A)
    raise ValueError(f"Cannot parse nucleus specification: {spec!r}")


def _nucleus_label(Z, A):
    """Mathtext legend label."""
    if Z == 0 and A == 1:
        return r'$n$'
    if Z == 1 and A == 1:
        return r'$p$'
    sym = _Z_TO_SYM.get(Z, f'Z{Z}')
    return rf'$^{{{A}}}{{\rm {sym}}}$'


# ── Main plotting function ────────────────────────────────────────────────────

def plot_abundance(sim, nuclei, quantity="Y", ax=None, **plot_kw):
    """Plot nuclear abundance trajectories vs. time.

    Parameters
    ----------
    sim : pandas.DataFrame
        Output of load_simulation(); MultiIndex (snap_idx, N, Z).
    nuclei : list
        Each element is a (Z, A) tuple or a string like 'Fe-56', 'n', 'p', 'He-4'.
    quantity : {'Y', 'X'}
        'Y' = number abundance per baryon; 'X' = mass fraction.
    ax : matplotlib.axes.Axes, optional
        Existing axes to draw on.
    **plot_kw
        Forwarded to ax.plot() for every curve.

    Returns
    -------
    matplotlib.axes.Axes
    """
    if quantity not in ('Y', 'X'):
        raise ValueError("quantity must be 'Y' or 'X'")
    if ax is None:
        _, ax = plt.subplots()

    flat = sim.reset_index()   # snap_idx, N, Z, A, Y, X, time, T9, rho

    # snapshot time index (snap_idx → time) used to fill missing nuclei with 0
    snap_times = (flat[['snap_idx', 'time']]
                  .drop_duplicates('snap_idx')
                  .sort_values('snap_idx'))

    for spec in nuclei:
        Z, A = _parse_nucleus(spec)
        N = A - Z
        label = _nucleus_label(Z, A)

        nuc = flat[(flat['N'] == N) & (flat['Z'] == Z)][['snap_idx', quantity]]
        merged = snap_times.merge(nuc, on='snap_idx', how='left')
        merged[quantity] = merged[quantity].fillna(0.0)

        times  = merged['time'].values
        values = merged[quantity].values
        # replace zero / negative with NaN so log-scale renders a gap, not a crash
        plot_vals = np.where(values > 0.0, values, np.nan)

        ax.plot(times, plot_vals, label=label, **plot_kw)

    ax.set_yscale('log')

    # auto-detect linear vs log x-axis from time dynamic range
    t_all = flat['time'].values
    t_pos = t_all[t_all > 0]
    if t_pos.size > 0 and (t_pos.max() / t_pos.min()) > 1e3:
        ax.set_xscale('log')
        ax.set_xlabel('Time [s]  (log scale)')
    else:
        ax.set_xlabel('Time [s]')

    ylabel = r'Number abundance $Y_i$' if quantity == 'Y' else r'Mass fraction $X_i$'
    ax.set_ylabel(ylabel)
    ax.legend(frameon=False)
    return ax
```

```python

# Load Run_1 — parses 5737 snapshot files; takes ~2-3 min first call, instant after (cached).
sim1 = load_simulation("1")

n_snaps = sim1.index.get_level_values('snap_idx').nunique()
t_min   = sim1['time'].min()
t_max   = sim1['time'].max()
print(f"Run_1: {n_snaps} snapshots, time range {t_min:.3e} – {t_max:.3e} s")
print(f"       {len(sim1):,} nucleus-snapshot entries")
```

```python

# Abundance trajectories for Run_1 — Y and X side by side
nuclei_plot = ['n', 'p', 'He-4', 'Ni-56', 'Ni-78']

fig, (ax_Y, ax_X) = plt.subplots(1, 2, figsize=(11, 4.5), dpi=150, sharey=False)

plot_abundance(sim1, nuclei_plot, quantity='Y', ax=ax_Y)
ax_Y.set_title('Number abundance $Y_i$')

plot_abundance(sim1, nuclei_plot, quantity='X', ax=ax_X)
ax_X.set_title('Mass fraction $X_i$')

fig.suptitle('Run_1 — nuclear abundance trajectories', fontsize=13)
fig.tight_layout()
plt.show()
```
